```python
def get_artist_id(name):
    try:
        results = sp.search(q='artist:' + name, type='artist', limit=1)
        items = results['artists']['items']
        if items:
            return items[0]['id'], items[0]['name']
        return None, None
    except:
        return None, None

ids, returned_names = [], []

for name in df['Artist']:
    spotify_id, returned_name = get_artist_id(name)
    ids.append(spotify_id)
    returned_names.append(returned_name)
    time.sleep(0.1)

df['spotify_id'] = ids
df['returned_name'] = returned_names
df.head()
```

```python
def get_genres(spotify_id):
    try:
        artist = sp.artist(spotify_id)
        return artist['genres']
    except:
        return []

genres = []

for spotify_id in df['spotify_id']:
    genre = get_genres(spotify_id)
    genres.append(genre)
    time.sleep(0.1)

df['genres'] = genres
df.head()
```

```python
import pandas as pd

coachella = pd.read_csv('coachella.csv')

tracks_df = pd.read_csv('coachella_headliners.csv')
tracks_df.head()
```

```python
features = [
    "Danceability", "Energy", "Key", "Loudness", "Mode",
    "Speechiness", "Acousticness", "Instrumentalness",
    "Liveness", "Valence", "Tempo", "Time Signature"
]

results = []
temp = {col: 0 for col in features}
count = 0

for i in range(len(tracks_df)):
    for col in features:
        temp[col] += tracks_df.loc[i, col]
    count += 1

    if i == len(tracks_df) - 1 or tracks_df.loc[i, "Album Name"] != tracks_df.loc[i + 1, "Album Name"]:
        row = {"Album Name": tracks_df.loc[i, "Album Name"], "Artist Name(s)": tracks_df.loc[i, "Artist Name(s)"]}
        for col in features:
            row[col] = round(temp[col] / count, 4)
        results.append(row)

        temp = {col: 0 for col in features}
        count = 0

output_df = pd.DataFrame(results)
output_df.head()

artists_df = pd.read_csv("coachella.csv")

merged = pd.concat([artists_df, output_df], axis=1)
merged.head()
```

```python
coachella_features_only = merged.drop(columns = ['Main?', 'spotify_id', 'Artist.1', 'Genre (specific)', 'Genre (broad)'])
coachella_features_only.to_csv('coachella_genres_features.csv', index=False)
coachella_features_only.tail()
```

```python
import pandas as pd
import plotly.graph_objects as go

# ── 1. FEATURE SELECTION ──────────────────────────────────────────────────────
FEATURE_1 = "valence"        # x-axis
FEATURE_2 = "acousticness"  # y-axis

# ── 2. LOAD & NORMALIZE DATA ──────────────────────────────────────────────────
FESTIVALS = {
    "coachella": {"file": "coachella_genres_features.csv", "color": "#ee9a05"},
    "bonnaroo":     {"file": "bonnaroo_genres_features.csv",     "color": "#B12011"},
    "lollapalooza": {"file": "lollapalooza_genres_features.csv", "color": "#079495"},
}

dfs = []
for name, cfg in FESTIVALS.items():
    d = pd.read_csv(cfg["file"])
    d.columns = d.columns.str.lower()  # ← add this line, normalizes ALL column names
    d = d.dropna(subset=[FEATURE_1, FEATURE_2, "year"])
    d["year"] = d["year"].astype(int)
    if "genre (broad)" in d.columns:
        d = d.rename(columns={"genre (broad)": "genre"})
    d["_festival"] = name
    dfs.append(d)

df_all = pd.concat(dfs, ignore_index=True)
years = sorted(df_all["year"].unique())

# ── 3. STYLING ────────────────────────────────────────────────────────────────
BG_COLOR   = "#F5EDD6"
PLOT_BG    = "#FFF9E7"
TEXT_COLOR = "#052D3A"
GRID_COLOR = "#D6C9A8"

# ── 4. SCATTER BUILDER ────────────────────────────────────────────────────────
def make_traces(year_df):
    traces = []
    for name, cfg in FESTIVALS.items():
        group = year_df[year_df["_festival"] == name]
        traces.append(go.Scatter(
            x=group[FEATURE_1] if len(group) > 0 else [],
            y=group[FEATURE_2] if len(group) > 0 else [],
            mode="markers",
            name=name,
            text=group["artist"] if len(group) > 0 else [],
            marker=dict(
                size=16,
                color=cfg["color"],
                line=dict(width=1.5, color=TEXT_COLOR)
            ),
            legendgroup=name,
            visible=True,
            hovertemplate=(
                "<b>%{text}</b><br>"
                f"{FEATURE_1}: %{{x:.2f}}<br>"
                f"{FEATURE_2}: %{{y:.2f}}<br>"
                f"Festival: {name}<extra></extra>"
            )
        ))
    return traces

# ── 5. BUILD FRAMES ───────────────────────────────────────────────────────────
frames = []
for year in years:
    year_df = df_all[df_all["year"] == year]
    frames.append(go.Frame(
        data=make_traces(year_df),
        layout=go.Layout(title_text=f"festival headliners — {year}"),
        name=str(year)
    ))

# ── 6. INITIAL FIGURE ────────────────────────────────────────────────────────
fig = go.Figure(
    data=make_traces(df_all[df_all["year"] == years[0]]),
    frames=frames
)

# ── 7. LAYOUT ────────────────────────────────────────────────────────────────
fig.update_layout(
    title=dict(
        text=f"festival headliners — {years[0]}",
        font=dict(size=22, color=TEXT_COLOR, family="Georgia, serif"),
        x=0.5
    ),
    xaxis=dict(
        title=FEATURE_1, range=[0, 1],
        gridcolor=GRID_COLOR, zerolinecolor=GRID_COLOR,
        tickfont=dict(color=TEXT_COLOR), titlefont=dict(color=TEXT_COLOR, size=18, family="Georgia, serif")
    ),
    yaxis=dict(
        title=FEATURE_2, range=[0, 1],
        gridcolor=GRID_COLOR, zerolinecolor=GRID_COLOR,
        tickfont=dict(color=TEXT_COLOR), titlefont=dict(color=TEXT_COLOR, size=18, family="Georgia, serif")
    ),
    plot_bgcolor=BG_COLOR,
    paper_bgcolor=PLOT_BG,
    font=dict(color=TEXT_COLOR),
    autosize=False,
    width=950,
    height=650,
    margin=dict(l=60, r=180, t=80, b=120),
    updatemenus=[dict(
        type="buttons", showactive=False,
        y=-0.12, x=0.5, xanchor="center",
        buttons=[
            dict(label="▶  play", method="animate",
                 args=[None, dict(frame=dict(duration=700, redraw=True), fromcurrent=True)]),
            dict(label="⏸  pause", method="animate",
                 args=[[None], dict(frame=dict(duration=0, redraw=False), mode="immediate")])
        ],
        font=dict(color=TEXT_COLOR), bgcolor="#E9D8A6", bordercolor=GRID_COLOR
    )],
    sliders=[dict(
        currentvalue=dict(prefix="year: ", font=dict(color=TEXT_COLOR, size=14)),
        pad=dict(t=40),
        steps=[dict(
            args=[[str(year)], dict(frame=dict(duration=800, redraw=True), mode="immediate")],
            label=str(year), method="animate"
        ) for year in years],
        font=dict(color=TEXT_COLOR), bgcolor="#E9D8A6",
        bordercolor=GRID_COLOR, tickcolor=TEXT_COLOR
    )]
)

fig.show()
fig.write_html("festival_headliners.html")
```

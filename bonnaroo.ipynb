**genre over time**

```python
import pandas as pd
import plotly.graph_objects as go
df = pd.read_csv('bonnaroo.csv')

colors = {
    'Rock':       '#8c181c',
    'Hip-hop':    '#cb6701',
    'Electronic': '#bb3e01',
    'Indie':      '#015f72',
    'Pop':        '#94d2bd',
    'R&B/Funk':   '#001218',
    'Country':    '#ee9a05',
    'Latin':      '#b12011',
    'K-pop':      '#bb3e01',
    'Other':      '#e9d8a6',
}

grouped = df.groupby(['year', 'Genre (broad)']).size().reset_index(name='count')
totals = grouped.groupby('year')['count'].transform('sum')
grouped['pct'] = (grouped['count'] / totals * 100).round(1)

pivot = grouped.pivot(index='year', columns='Genre (broad)', values='pct').fillna(0)
pivot = pivot.sort_index()

fig = go.Figure()

for genre in colors:
    if genre in pivot.columns:
        fig.add_trace(go.Scatter(
            x=pivot.index,
            y=pivot[genre],
            name=genre,
            stackgroup='one',
            fillcolor=colors[genre],
            line=dict(color=colors[genre]),
            hovertemplate=f'%{{x}}<br>{genre}: %{{y:.1f}}%<extra></extra>'
        ))

BG_COLOR    = "#FFF9E7"
TEXT_COLOR = "#052D3A"
fig.update_layout(
    title=dict(
        text='bonnaroo genre over time',
        x=0.5,
        font=dict(size=22, color=TEXT_COLOR, family="Georgia, serif")
        ),
    height=500,
    width=900,
    hovermode='x unified',
    legend=dict(orientation='h', yanchor='bottom', y=-0.3),
    plot_bgcolor='white',
    xaxis=dict(
        title='year',
        tickfont=dict(color=TEXT_COLOR), titlefont=dict(color=TEXT_COLOR, size=18, family="Georgia, serif")
    ),
    yaxis=dict(
        title='% of headliners', range=[0, 100],
        tickfont=dict(color=TEXT_COLOR), titlefont=dict(color=TEXT_COLOR, size=18, family="Georgia, serif")
    ),
    paper_bgcolor=BG_COLOR,
)

fig.show()
fig.write_html("bonnaroo_genres.html")
```

**growth rates**

```python
import pandas as pd
import plotly.graph_objects as go

df = pd.read_csv("bonnaroo growth rates.csv", index_col=0)

x = df.loc["announce"].astype(float).values / 1e6
y = df.loc["% change"].astype(float).values * 100
labels = df.columns.tolist()
genres = df.loc["genre"].values.tolist()

colors_map = {
    'Rock':       '#8c181c',
    'Hip-hop':    '#cb6701',
    'Electronic': '#bb3e01',
    'Indie':      '#015f72',
    'Pop':        '#94d2bd',
    'R&B/Funk':   '#001218',
    'Country':    '#ee9a05',
    'Latin':      '#b12011',
    'K-pop':      '#bb3e01',
    'Other':      '#e9d8a6',
}

colors = [colors_map.get(g, "#999999") for g in genres]

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x, y=y,
    mode="markers+text",
    text=labels,
    textposition="top right",
    marker=dict(size=14, color=colors, line=dict(width=1.5, color="#052D3A")),
    hovertemplate="<b>%{text}</b><br>Listeners: %{x:.1f}M<br>Growth: %{y:.1f}%<extra></extra>"
))

fig.add_hline(y=0, line=dict(color="black", width=2))

fig.update_layout(
    plot_bgcolor="#F5EDD6",
    paper_bgcolor="#FFF9E7",
    xaxis=dict(
        title=dict(
            text="listeners at announcement (millions)",
            font=dict(size=18, color="#052D3A", family="Georgia, serif")
        ),
        gridcolor="#D6C9A8"
    ),
    yaxis=dict(
        title=dict(
            text="6-month growth (%)",
            font=dict(size=18, color="#052D3A", family="Georgia, serif")
        ),
        gridcolor="#D6C9A8"
    ),
    title=dict(
        text="bonnaroo headliner growth rates",
        x=0.5,
        font=dict(size=22, color="#052D3A", family="Georgia, serif")
    ),
    height=600,
    width=950,
)

fig.show()
fig.write_html("bonnaroo_growth.html")
```
